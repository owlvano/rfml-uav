```python
import os
import numpy as np
import matplotlib.pyplot as plt
from dotenv import load_dotenv

from rfml_uav.drone_rf.consts import BUI_dict as BUI

load_dotenv()
```

```python
# Path of raw RF data
filename = os.path.join(os.getenv("DATA_PATH"), "raw")
```

```python
drone_type = "AR"
mode = 0
bui = BUI[drone_type][mode]
n = 1

x = np.loadtxt(os.path.join(filename, f"{bui}L_{n}.csv"), delimiter=',')
y = np.loadtxt(os.path.join(filename, f"{bui}H_{n}.csv"), delimiter=',')

# Normalize
# x = x / np.max(np.abs(x))
# y = y / np.max(np.abs(y))
```

```python
# Time axis
t = np.arange(0, len(x))

# Plotting
plt.figure(figsize=(8.4, 6))  # size in inches (840x600 pixels at 100 dpi)
plt.plot(t, x, label=r'$x^{(L)}$')
plt.plot(t, y, 'r', label=r'$x^{(H)}$')
plt.grid(True, which='both')
plt.xlabel('Time (s)', fontsize=18)
plt.legend(loc='lower left')
plt.xticks(fontweight='bold', fontsize=20)
plt.title(f"Drone: {drone_type}, mode: {mode}, segment {n}")
plt.tight_layout()
```
