```python
import os
import numpy as np
from dotenv import load_dotenv

from rfml_uav.drone_rf.consts import BUI

load_dotenv()
```

```python
N_seg = 1e7
filename = os.path.join(os.getenv("DATA_PATH"), "raw")  # Path of raw RF data
```

```python
# Counting
c = np.zeros(len(BUI), dtype=int)
for i in range(len(BUI)):
    c[i] = len([f for f in os.listdir(filename) if f.startswith(BUI[i])])

# Level 1 details
L1 = np.zeros(2)
L1[0] = N_seg * np.sum(c[1:])
L1[1] = N_seg * c[0]
P1 = 100 * L1 / np.sum(L1)

# Level 2 details
L2 = np.zeros(4)
L2[0] = N_seg * np.sum(c[1:5])
L2[1] = N_seg * np.sum(c[5:9])
L2[2] = N_seg * c[9]
L2[3] = N_seg * c[0]
P2 = 100 * L2 / np.sum(L2)

# Level 3 details
L3 = np.zeros(10)
L3[0] = N_seg * c[1]
L3[1] = N_seg * c[2]
L3[2] = N_seg * c[3]
L3[3] = N_seg * c[4]
L3[4] = N_seg * c[5]
L3[5] = N_seg * c[6]
L3[6] = N_seg * c[7]
L3[7] = N_seg * c[8]
L3[8] = N_seg * c[9]
L3[9] = N_seg * c[0]
P3 = 100 * L3 / np.sum(L3)

# Display
print(f"L2 details: {L2/N_seg/2} {L2} {P2}")
```
